```python
base_url = "https://raw.githubusercontent.com/MiguelSteph/reinforcement_learning_algorithms_with_jax/ppo/"

def download_file_from_github(file_path: str, file_name: str):
    if not os.path.isfile(file_name):
        file_url = base_url + file_path
        print(f"Downloading {file_url}...")
        try:
            urllib.request.urlretrieve(file_url, file_name)
        except HTTPError as e:
            print("Something went wrong. Please try to download the file directly from the GitHub repository:\n", e)


file_paths = [
    'ppo/configs.py',
    'ppo/ppo_types.py',
    'ppo/model.py',
    'ppo/trainer.py',
    'ppo/agent.py',
    'ppo/envs_wrapper.py',
]
file_names = [
    'configs.py',
    'ppo_types.py',
    'model.py',
    'trainer.py',
    'agent.py',
    'envs_wrapper.py',
]

for file_path, file_name in zip(file_paths, file_names): 
    download_file_from_github(file_path, file_name)
```

```python
from ppo_types import NetworkConfig, TrainerConfig
from agent import Agent
from trainer import AgentTrainer
import jax
```

```python
network_config = NetworkConfig()
trainer_config = TrainerConfig()

initial_rng_key = jax.random.key(18)
agent_rng_key, trainer_rng_key = jax.random.split(initial_rng_key)

agent = Agent(network_config)
state = agent.init(agent_rng_key, trainer_config.obs_shape, trainer_config.learning_rate)

trainer = AgentTrainer(agent, trainer_config, trainer_rng_key)
state = trainer.run_train_loop(state)
```
