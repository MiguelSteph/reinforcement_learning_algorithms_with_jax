```python
base_url = "https://raw.githubusercontent.com/MiguelSteph/reinforcement_learning_algorithms_with_jax/main/"

def download_file_from_github(file_path: str, file_name: str):
    if not os.path.isfile(file_name):
        file_url = base_url + file_path
        print(f"Downloading {file_url}...")
        try:
            urllib.request.urlretrieve(file_url, file_name)
        except HTTPError as e:
            print("Something went wrong. Please try to download the file directly from the GitHub repository:\n", e)


file_paths = [
    'dqn_with_vanilla_replay_buffer/configs.py',
    'dqn_with_vanilla_replay_buffer/dqn_types.py',
    'dqn_with_vanilla_replay_buffer/model.py',
    'dqn_with_vanilla_replay_buffer/replay_buffer.py',
    'dqn_with_vanilla_replay_buffer/trainer.py',
    'dqn_with_vanilla_replay_buffer/agent.py',
    'dqn_with_vanilla_replay_buffer/env_wrapper.py',
]
file_names = [
    'configs.py',
    'dqn_types.py',
    'model.py',
    'replay_buffer.py',
    'trainer.py',
    'agent.py',
    'env_wrapper.py',
]

for file_path, file_name in zip(file_paths, file_names): 
    download_file_from_github(file_path, file_name)
```

```python
from dqn_types import DQNConfig, TrainerConfig
from agent import Agent
from trainer import AgentTrainer
from env_wrapper import EnvWrapper
import jax
```

```python
env = EnvWrapper()
# eval_env = EnvWrapper(terminal_on_life_loss=False)
dqn_config = DQNConfig()
trainer_config = TrainerConfig()

# Create the agent
agent = Agent(dqn_config)
agent_init_rng = jax.random.key(20)

dqn_state = agent.init(
    agent_init_rng, 
    trainer_config.obs_shape, 
    trainer_config.learning_rate
)

# Create the agent trainer
trainer_initial_rng = jax.random.key(18)
trainer = AgentTrainer(agent, trainer_config, trainer_initial_rng)
dqn_state = trainer.run_train_loop(env, dqn_state)
```
